```python

import pandas as pd
import matplotlib.pyplot as plt



df = pd.read_csv("air_quality.csv")

print("Dataset loaded successfully!")
```

## Q1. Dataset Loading and Basic Exploration

```python
# First 5 records
print("First 5 records:")
print(df.head())

# Last 5 records
print("\nLast 5 records:")
print(df.tail())

# Number of rows and columns
print("\nNumber of rows and columns:")
print(df.shape)

# Column names and data types
print("\nColumn names:")
print(df.columns)

print("\nData types:")
print(df.dtypes)
```

## Q2. Descriptive Analysis

```python
# Basic statistics for AQI and PM2.5

print("AQI")
print("Mean:", df["AQI"].mean())
print("Median:", df["AQI"].median())
print("Minimum:", df["AQI"].min())
print("Maximum:", df["AQI"].max())
print("Standard Deviation:", df["AQI"].std())

print("\nPM2.5")
print("Mean:", df["PM2.5"].mean())
print("Median:", df["PM2.5"].median())
print("Minimum:", df["PM2.5"].min())
print("Maximum:", df["PM2.5"].max())
print("Standard Deviation:", df["PM2.5"].std())
```

## Q3. Missing Value Handling

```python
# Count missing values in each column
print("Missing values in each column:")
print(df.isnull().sum())

# Fill missing numerical values with the column mean
df["AQI"] = df["AQI"].fillna(df["AQI"].mean())
df["PM2.5"] = df["PM2.5"].fillna(df["PM2.5"].mean())

# Fill missing city names with "Unknown"
df["City"] = df["City"].fillna("Unknown")

print("\nDataset after missing value treatment:")
print(df)
```

## Q4. Data Filtering and Queries

```python
# a) Records where AQI > 200
print("a) AQI > 200:")
print(df[df["AQI"] > 200])

# b) Records where PM2.5 > 100 and AQI > 200
print("\nb) PM2.5 > 100 and AQI > 200:")
print(df[(df["PM2.5"] > 100) & (df["AQI"] > 200)])

# c) Records of one selected city
selected_city = "Ahmedabad"   # Change city name if required
print("\nc) Records of", selected_city, ":")
print(df[df["City"] == selected_city])
```

## Q5. Sorting and Top Records

```python
# Sort by AQI in descending order
sorted_df = df.sort_values("AQI", ascending=False)

# Display 10 records with highest AQI
print("Top 10 records with highest AQI:")
print(sorted_df.head(10))
```

## Q6. Group-wise Analysis

```python
# Average AQI for each city
average_aqi = df.groupby("City")["AQI"].mean()

print("a) Average AQI for each city:")
print(average_aqi)

# Maximum AQI for each city
maximum_aqi = df.groupby("City")["AQI"].max()

print("\nb) Maximum AQI for each city:")
print(maximum_aqi)

# Top 5 cities based on average AQI
top_5_cities = average_aqi.sort_values(ascending=False).head(5)

print("\nc) Top 5 cities based on average AQI:")
print(top_5_cities)
```

## Q7. Outlier Detection – IQR Method

```python
# Find Q1 and Q3 for PM2.5
Q1 = df["PM2.5"].quantile(0.25)
Q3 = df["PM2.5"].quantile(0.75)

# Calculate IQR
IQR = Q3 - Q1

# Calculate lower and upper limits
lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

# Find outliers
outliers = df[(df["PM2.5"] < lower_limit) | (df["PM2.5"] > upper_limit)]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower limit:", lower_limit)
print("Upper limit:", upper_limit)
print("Number of outliers:", len(outliers))
```

## Q8. Outlier Visualization

```python
# Box plot for PM2.5 and AQI
plt.figure(figsize=(7, 5))
plt.boxplot([df["PM2.5"], df["AQI"]], labels=["PM2.5", "AQI"])
plt.title("Box Plot of PM2.5 and AQI")
plt.ylabel("Value")
plt.show()

# Observations:
print("Observation 1: Points outside the whiskers indicate possible outliers.")
print("Observation 2: The box plot also shows the spread and median of PM2.5 and AQI.")
```

## Q9. Data Visualization

```python
# a) Histogram of AQI
plt.figure(figsize=(7, 5))
plt.hist(df["AQI"], bins=10)
plt.title("Distribution of AQI")
plt.xlabel("AQI")
plt.ylabel("Frequency")
plt.show()

# b) Bar chart of average AQI for top 10 cities
top_10_cities = df.groupby("City")["AQI"].mean().sort_values(ascending=False).head(10)

plt.figure(figsize=(9, 5))
plt.bar(top_10_cities.index, top_10_cities.values)
plt.title("Average AQI of Top 10 Cities")
plt.xlabel("City")
plt.ylabel("Average AQI")
plt.xticks(rotation=45)
plt.show()
```

## Q10. Relationship Analysis

```python
# Scatter plot between PM2.5 and AQI
plt.figure(figsize=(7, 5))
plt.scatter(df["PM2.5"], df["AQI"])
plt.title("Relationship Between PM2.5 and AQI")
plt.xlabel("PM2.5")
plt.ylabel("AQI")
plt.show()

# Correlation
correlation = df["PM2.5"].corr(df["AQI"])

print("Correlation between PM2.5 and AQI:", correlation)

# Simple conclusion
if correlation > 0:
    print("Conclusion: PM2.5 and AQI have a positive relationship.")
elif correlation < 0:
    print("Conclusion: PM2.5 and AQI have a negative relationship.")
else:
    print("Conclusion: PM2.5 and AQI have no linear relationship.")
```

## Quick Viva Revision

- **AQI:** Air Quality Index.
- **PM2.5:** Fine particulate matter with diameter of about 2.5 micrometres or smaller.
- **Mean:** Average value.
- **Median:** Middle value after sorting.
- **IQR:** Q3 − Q1.
- **Outlier by IQR:** A value below Q1 − 1.5×IQR or above Q3 + 1.5×IQR.
- **Correlation:** Measures the strength and direction of a linear relationship between two variables.
- **Histogram:** Shows the distribution of a numerical variable.
- **Box plot:** Shows median, spread and possible outliers.
- **GroupBy:** Used to perform calculations separately for each group, such as each city.
